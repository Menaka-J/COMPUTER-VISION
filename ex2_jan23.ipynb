```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
ig = cv2.imread(r"/content/stranger things ig.jpg")
ig = cv2.cvtColor(ig, cv2.COLOR_BGR2RGB)
```

# ***RGB TO CMY***

```python
r,g,b = ig[:,:,0], ig[:,:,1], ig[:,:,2]
r,g,b = r/255, g/255, b/255
```

```python
c = 1 - r
m = 1 - g
y = 1 - b

cmyig = (np.dstack((c,m,y))*255).astype(np.uint8)
```

```python
plt.subplot(1,2,1)
plt.imshow(ig)
plt.title("Original Image")

plt.subplot(1,2,2)
plt.imshow(cmyig)
plt.title("CMY Image")

plt.savefig("cmy-image1.jpg")
```

# **RGB TO HSV**

```python
hsvig = cv2.cvtColor(ig, cv2.COLOR_RGB2HSV)
```

```python
plt.subplot(1,2,1)
plt.imshow(ig)
plt.title("Original Image")

plt.subplot(1,2,2)
plt.imshow(hsvig)
plt.title("HSV Image")

plt.savefig("hsv-image1.jpg")
```

# ***RGB TO YIQ***

```python
y = 0.299*r + 0.587*g + 0.114*b
i = 0.596*r - 0.274*g - 0.321*b
q = 0.211*r- 0.522*g + 0.311*b
yiqig = (np.dstack((y,i,q))*255).astype(np.uint8)
```

```python
plt.subplot(1,2,1)
plt.imshow(ig)
plt.title("Original Image")

plt.subplot(1,2,2)
plt.imshow(yiqig)
plt.title("YIQ Image")

plt.savefig("yiq-image1.jpg")
```

# ***SHOWING SEPARATE RGB PLANES***

```python
row, col = r.shape
r_plane = np.zeros((row, col, 3), dtype = np.uint8)
g_plane = np.zeros((row, col, 3), dtype = np.uint8)
b_plane = np.zeros((row, col, 3), dtype = np.uint8)

r_plane[:,:,0] = r*255
g_plane[:,:,1] = g*255
b_plane[:,:,2] = b*255

plt.figure(figsize = (12,4))

plt.subplot(1,3,1)
plt.imshow(r_plane)
plt.title("Red Plane")

plt.subplot(1,3,2)
plt.imshow(g_plane)
plt.title("Green Plane")

plt.subplot(1,3,3)
plt.imshow(b_plane)
plt.title("Blue Plane")

plt.savefig("rgb-planes1-image1.jpg")
```

# ***SEPARATING HSV CHANNEL***

```python
h, s, v = cv2.split(hsvig)

plt.figure(figsize = (12,4))

plt.subplot(1,3,1)
plt.imshow(h)
plt.title("Hue Channel")

plt.subplot(1,3,2)
plt.imshow(s, cmap = 'grey')
plt.title("Saturation Channel")

plt.subplot(1,3,3)
plt.imshow(v, cmap = 'grey')
plt.title("Value Channel")

plt.savefig("hue-channel1-image1.jpg")
```

# ***ALTERING HSV VALUES***

```python
h = cv2.cvtColor(ig, cv2.COLOR_RGB2HSV)
h1 = cv2.cvtColor(ig, cv2.COLOR_RGB2HSV)
h2 = cv2.cvtColor(ig, cv2.COLOR_RGB2HSV)
h3 = cv2.cvtColor(ig, cv2.COLOR_RGB2HSV)
```

```python
h1[:,:,1]=150
h2[:,:,2]=200
h3[:,:,0]=180
```

```python
plt.figure(figsize = (12, 8))

plt.subplot(2,2,1)
plt.title('HSV image')
plt.imshow(hsvig)

plt.subplot(2,2,2)
plt.title('SATURATION ALTERED')
plt.imshow(h1, cmap = 'grey')

plt.subplot(2,2,3)
plt.title('VALUE ALTERED')
plt.imshow(h2, cmap = 'grey')

plt.subplot(2,2,4)
plt.title('HUE ALTERED')
plt.imshow(h3)

plt.savefig("hsv-altered1-image1.jpg")
```

# ***COLOR MODEL CREATION***

```python
row, col = 100,100
newig = np.zeros((row, col, 3), dtype = np.uint8)

for i in range(0, row, 25):
    for j in range(0, col, 25):
        newig[i:i+25, j:j+25, 0] = np.random.randint(0, 256)
        newig[i:i+25, j:j+25, 1] = np.random.randint(0, 256)
        newig[i:i+25, j:j+25, 2] = np.random.randint(0, 256)

plt.imshow(newig)
plt.title("New Color Model")
plt.savefig("cm1.jpg")
```
