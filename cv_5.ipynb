```python
import cv2 
import numpy as np 
import matplotlib.pyplot as plt 
ig = cv2.imread(r"C:\Users\admin\Pictures\computer vision s4\ben stokes.jpg") 
ig = cv2.cvtColor(ig, cv2.COLOR_BGR2GRAY) 
 
sobalymat = np.array([[1,1,1],[0,0,0],[-1,-1,-1]]) 
sobalxmat = np.array([[-1,0,1],[-1,0,1],[-1,0,1]]) 

sobalx2mat = np.array([[1,2,3,2,1],[2,3,5,3,2],[0,0,0,0,0],[-2,-3,-5,-3,-2],[-1,-2,-3,-2,-1]]) 
sobaly2mat = np.array([[-1,-1,0,2,1],[-2,-3,0,3,2],[-3,-5,0,5,3],[-2,-3,0,3,2],[-1,-2,0,2,1]]) 

row,col = ig.shape 
sobalx = np.zeros_like(ig, dtype = np.float32) 
sobaly = np.zeros_like(ig, dtype = np.float32) 

for i in range(1,row-1): 
    for j in range(1,col-1): 
 
        kernel = ig[i-1:i+2, j-1:j+2] 
 
        sobalx[i,j] = np.sum(kernel * sobalxmat) 
        sobaly[i,j] = np.sum(kernel * sobalymat) 
 
mag = np.sqrt((sobalx**2) + (sobaly**2)) 
mag = np.uint8(255 * mag/ (np.max(mag) - np.min(mag))) 

plt.subplot(1,2,1) 
plt.imshow(ig, cmap = 'grey') 
plt.title("Original Image") 
plt.axis('off') 
 
plt.subplot(1,2,2) 
plt.imshow(mag, cmap = 'grey') 
plt.title("Sobel edge detected Image") 
plt.axis('off') 
```

```python
def add_salt_and_pepper_noise(img, prob): 
    noisy_img = img.copy() 
    noise = np.random.rand(*img.shape) 
    noisy_img[noise < prob / 2] = 0 
    noisy_img[noise > 1 - prob / 2] = 255 
    return noisy_img 
    
noisy_image = add_salt_and_pepper_noise(ig, 0.5) 
denoised_image = cv2.medianBlur(noisy_image, 5) 

plt.figure(figsize=(12, 4)) 

plt.subplot(2, 3, 1) 
plt.imshow(ig, cmap='gray') 
plt.title('Original Image') 
plt.axis('off') 

plt.subplot(2, 3, 2) 
plt.imshow(noisy_image, cmap='gray') 
plt.title('salt&pepper noise Image') 
plt.axis('off') 

plt.subplot(2, 3, 3) 
plt.imshow(denoised_image, cmap='gray') 
plt.title('Denoised Image') 
plt.axis('off') 

plt.subplot(2,3,4) 
plt.hist(ig.ravel(),256,[0,256]) 

plt.subplot(2,3,5) 
plt.hist(noisy_image.ravel(),256,[0,256]) 

plt.subplot(2,3,6) 
plt.hist(denoised_image.ravel(),256,[0,256]) 

plt.show() 
```

```python
def add_white_impulse_noise(img, prob): 
    noisy_img = img.copy() 
    noise = np.random.rand(*img.shape) 
    noisy_img[noise < prob] = 255 
    return noisy_img 
    
noisy_image = add_white_impulse_noise(ig, 0.2) 
denoised_image = cv2.medianBlur(noisy_image, 5) 

plt.figure(figsize=(12, 4)) 

plt.subplot(2, 3, 1) 
plt.imshow(ig, cmap='gray') 
plt.title("Original Image") 
plt.axis("off") 

plt.subplot(2, 3, 2) 
plt.imshow(noisy_image, cmap='gray') 
plt.title("Impulse Noise") 
plt.axis("off") 

plt.subplot(2, 3, 3) 
plt.imshow(denoised_image, cmap='gray') 
plt.title("Denoised Image") 
plt.axis("off") 

plt.subplot(2,3,4) 
plt.hist(ig.ravel(),256,[0,256]) 

plt.subplot(2,3,5) 
plt.hist(noisy_image.ravel(),256,[0,256]) 

plt.subplot(2,3,6) 
plt.hist(denoised_image.ravel(),256,[0,256]) 

plt.show() 
```

```python
def add_gaussian_noise(img, mean=10, stddev=30): 
    noise = np.random.normal(mean, stddev, img.shape).astype(np.int16) 
    noisy_img = np.clip(img.astype(np.int16) + noise, 0, 255).astype(np.uint8) 
    return noisy_img 
    
noisy_image = add_gaussian_noise(ig) 
denoised_image = cv2.GaussianBlur(noisy_image, (5,5), 2)

plt.figure(figsize=(12, 4)) 

plt.subplot(2, 3, 1) 
plt.imshow(ig, cmap='gray') 
plt.title("Original Image") 
plt.axis("off") 

plt.subplot(2, 3, 2) 
plt.imshow(noisy_image, cmap='gray') 
plt.title("Gaussian Noise") 
plt.axis("off") 

plt.subplot(2, 3, 3) 
plt.imshow(denoised_image, cmap='gray') 
plt.title("Denoised Image") 
plt.axis("off") 

plt.subplot(2,3,4) 
plt.hist(ig.ravel(),256,[0,256]) 

plt.subplot(2,3,5) 
plt.hist(noisy_image.ravel(),256,[0,256]) 

plt.subplot(2,3,6) 
plt.hist(denoised_image.ravel(),256,[0,256]) 

plt.show() 
```

```python
def add_uniform_noise(img, low=-70, high=70): 
    noise = np.random.uniform(low, high, img.shape).astype(np.int16) 
    noisy_img = np.clip(img.astype(np.int16) + noise, 0, 255).astype(np.uint8) 
    return noisy_img 
    
noisy_image = add_uniform_noise(ig) 
denoised_image = cv2.medianBlur(noisy_image, 5) 

plt.figure(figsize=(12, 4)) 

plt.subplot(2, 3, 1) 
plt.imshow(ig, cmap='gray') 
plt.title("Original Image") 
plt.axis("off") 

plt.subplot(2, 3, 2) 
plt.imshow(noisy_image, cmap='gray') 
plt.title("Uniform Noise") 
plt.axis("off") 

plt.subplot(2, 3, 3) 
plt.imshow(denoised_image, cmap='gray') 
plt.title("Denoised Image") 
plt.axis("off") 

plt.subplot(2,3,4) 
plt.hist(ig.ravel(),256,[0,256]) 

plt.subplot(2,3,5) 
plt.hist(noisy_image.ravel(),256,[0,256]) 

plt.subplot(2,3,6) 
plt.hist(denoised_image.ravel(),256,[0,256]) 

plt.show()
```

```python
def add_rayleigh_noise(image, scale=10): 
    noise = np.random.rayleigh(scale, size=image.shape) 
    noisy_image = image + noise 
    noisy_image = np.clip(noisy_image, 0, 255).astype(np.uint8) 
    return noisy_image 
    
def restore_rayleigh_noise(noisy_image, original_image, scale=10): 
    restored_image = noisy_image - np.random.rayleigh(scale, size=noisy_image.shape) 
    restored_image = np.clip(restored_image, 0, 255).astype(np.uint8) 
    return restored_image 
    
noisy_image = add_rayleigh_noise(ig) 
denoised_image = restore_rayleigh_noise(noisy_image, ig) 

plt.figure(figsize=(12, 4)) 

plt.subplot(2, 3, 1) 
plt.imshow(ig, cmap='gray') 
plt.title("Original Image") 
plt.axis("off") 

plt.subplot(2, 3, 2) 
plt.imshow(noisy_image, cmap='gray') 
plt.title("Rayleigh Noise") 
plt.axis("off") 

plt.subplot(2, 3, 3) 
plt.imshow(denoised_image, cmap='gray') 
plt.title("Denoised Image") 
plt.axis("off") 

plt.subplot(2,3,4) 
plt.hist(ig.ravel(),256,[0,256]) 

plt.subplot(2,3,5) 
plt.hist(noisy_image.ravel(),256,[0,256]) 

plt.subplot(2,3,6) 
plt.hist(denoised_image.ravel(),256,[0,256]) 

plt.show() 
 
```

```python
def add_exponential_noise(img, lambda_val=0.05): 
    scale = 1 / lambda_val 
    noise = np.random.exponential(scale, img.shape).astype(np.uint8) 
    noisy_img = np.clip(img + noise, 0, 255).astype(np.uint8) 
    return noisy_img 
    
lambda_value = 0.05 
noisy_image = add_exponential_noise(ig, lambda_value) 
denoised_image = cv2.medianBlur(noisy_image, 5) 

plt.figure(figsize=(12, 4)) 

plt.subplot(2, 3, 1) 
plt.imshow(ig, cmap='gray') 
plt.title("Original Image") 
plt.axis("off") 

plt.subplot(2, 3, 2) 
plt.imshow(noisy_image, cmap='gray') 
plt.title("Exponential Noise") 
plt.axis("off") 

plt.subplot(2, 3, 3) 
plt.imshow(denoised_image, cmap='gray') 
plt.title("Denoised Image") 
plt.axis("off") 

plt.subplot(2,3,4) 
plt.hist(ig.ravel(),256,[0,256]) 

plt.subplot(2,3,5) 
plt.hist(noisy_image.ravel(),256,[0,256]) 

plt.subplot(2,3,6) 
plt.hist(denoised_image.ravel(),256,[0,256]) 

plt.show() 
```
