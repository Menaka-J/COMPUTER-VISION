```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
```

```python
ig = cv2.imread(r"/content/got.jpg")
```

```python
gray = cv2.cvtColor(ig, cv2.COLOR_BGR2GRAY)
```

```python
fourier = cv2.dft(np.float32(gray), flags=cv2.DFT_COMPLEX_OUTPUT)
fourier_shift = np.fft.fftshift(fourier)
magnitude = 20 * np.log(cv2.magnitude(fourier_shift[:, :, 0], fourier_shift[:, :, 1]))

magnitude = cv2.normalize(magnitude, None, 0, 255, cv2.NORM_MINMAX, cv2.CV_8UC1)

plt.figure(figsize=(10, 6))

plt.subplot(2, 2, 1)
plt.imshow(gray, cmap='grey')
plt.title("Original Image")
plt.axis('off')

plt.subplot(2, 2, 2)
plt.imshow(magnitude, cmap='gray')
plt.title("Fourier Transform Image")
plt.axis('off')

plt.subplot(2, 2, 3)
plt.hist(gray)

plt.subplot(2, 2, 4)
plt.hist(magnitude)

plt.show()
```

blur

```python
height, width = gray.shape

blurred_image = np.zeros_like(gray)

kernel_size = 7

kernel_radius = kernel_size // 2

for i in range(kernel_radius, height - kernel_radius):
    for j in range(kernel_radius, width - kernel_radius):

        region = gray[i - kernel_radius:i + kernel_radius + 1, j - kernel_radius:j + kernel_radius + 1]

        blurred_image[i, j] = np.mean(region)

plt.figure(figsize=(10, 5))

plt.subplot(2, 2, 1)
plt.imshow(gray, cmap = 'grey')
plt.title('Original Image')
plt.axis('off')

plt.subplot(2, 2, 2)
plt.imshow(blurred_image, cmap = 'grey')
plt.title('Blurred Image')
plt.axis('off')

plt.subplot(2, 2, 3)
plt.hist(gray)

plt.subplot(2, 2, 4)
plt.hist(blurred_image)
plt.show()
```

sharpening

```python
k= np.array([[0, -1, 0], [-1, 5, -1], [0, -1, 0]])
s= cv2.filter2D(gray, -1, k)

plt.subplot(2,2,1)
plt.imshow(gray,cmap='grey')
plt.title("Original Image")
plt.axis('off')

plt.subplot(2,2,2)
plt.imshow(s,cmap='grey')
plt.title("Sharpened Image")
plt.axis('off')

plt.subplot(2,2,3)
plt.hist(gray)
plt.xlim(0, 255)


plt.subplot(2,2,4)
plt.hist(s)
plt.xlim(0, 255)
```

```python

i2 = cv2.add(gray, 100)

plt.subplot(221)
plt.imshow(gray, cmap='gray')
plt.axis('off')
plt.title("Original Image")

plt.subplot(222)
plt.imshow(i2, cmap='gray')
plt.axis('off')
plt.title("Image + 100")

plt.subplot(223)
plt.hist(gray)
plt.xlim(0, 255)

plt.subplot(224)
plt.hist(i2)
plt.xlim(0, 255)

plt.tight_layout()
plt.show()
```

```python
i3=cv2.add(gray,-100)

plt.subplot(2,2,1)
plt.imshow(gray,cmap='grey')
plt.axis('off')
plt.title("Original Image")

plt.subplot(2,2,2)
plt.axis('off')
plt.imshow(i3,cmap='grey')
plt.title("Image - 100")

plt.subplot(2,2,3)
plt.hist(gray)
plt.xlim(0, 255)

plt.subplot(2,2,4)
plt.hist(i3)
plt.xlim(0, 255)
```

```python
r,c=gray.shape
x=np.zeros_like(gray)
for i in range(r):
    for j in range(c):
        x[i][j]=255-gray[i][j]


plt.subplot(2,2,1)
plt.imshow(gray,cmap='grey')
plt.axis('off')
plt.title("Original Image")

plt.subplot(2,2,2)
plt.axis('off')
plt.imshow(x,cmap='grey')
plt.title("Negative Image")

plt.subplot(2,2,3)
plt.hist(gray)

plt.subplot(2,2,4)
plt.hist(x)
```

```python
x1=np.zeros_like(gray)
x1=255+((gray-np.min(gray))/(np.max(gray)-np.min(gray)))

plt.subplot(2,2,1)
plt.imshow(gray,cmap='grey')
plt.axis('off')
plt.title("Original Image")

plt.subplot(2,2,2)
plt.axis('off')
plt.imshow(x1,cmap='grey')
plt.title("Contrast Image")

plt.subplot(2,2,3)
plt.hist(gray)

plt.subplot(2,2,4)
plt.hist(x1)
```

```python
def thr(y, n):
    r,c = y.shape

    for i in range(r):
        for j in range(c):
            if y[i, j] > n:
                y[i, j] = 0
            else:
                y[i, j] = 255
    return y

t=thr(gray,80)
gray = cv2.cvtColor(ig, cv2.COLOR_BGR2GRAY)

plt.subplot(221)
plt.imshow(gray,cmap='grey')
plt.axis('off')
plt.title("Original Image")

plt.subplot(222)
plt.axis('off')
plt.imshow(t,cmap='grey')
plt.title("Threshold Image")

plt.subplot(223)
plt.hist(gray)

plt.subplot(2,2,4)
plt.hist(t)
```
